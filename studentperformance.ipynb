```
import pandas as pd
import matplotlib.pyplot as plt
import zipfile
import io
import openpyxl

# Loading datasets from ZIP file

zip_file = zipfile.ZipFile("archive (1).zip")

maths = pd.read_excel(io.BytesIO(zip_file.read("Maths.csv")))
portuguese = pd.read_excel(io.BytesIO(zip_file.read("Portuguese.csv")))


# 1. Basic information

print("MATHEMATICS DATASET")
print(maths.head())

print("\nPORTUGUESE DATASET")
print(portuguese.head())

print("\nMaths Dataset Information")
print(maths.info())

print("\nPortuguese Dataset Information")
print(portuguese.info())


# 2. Average grades

maths_average = maths["G3"].mean()
portuguese_average = portuguese["G3"].mean()

print("\nAverage Mathematics Grade:", maths_average)
print("Average Portuguese Grade:", portuguese_average)


# 3. Highest and lowest grades

print("\nHighest Mathematics Grade:", maths["G3"].max())
print("Lowest Mathematics Grade:", maths["G3"].min())

print("\nHighest Portuguese Grade:", portuguese["G3"].max())
print("Lowest Portuguese Grade:", portuguese["G3"].min())


# 4. Students who passed and failed

maths_passed = maths[maths["G3"] >= 10]
maths_failed = maths[maths["G3"] < 10]

print("\nMathematics Results")
print("Passed:", len(maths_passed))
print("Failed:", len(maths_failed))


port_passed = portuguese[portuguese["G3"] >= 10]
port_failed = portuguese[portuguese["G3"] < 10]

print("\nPortuguese Results")
print("Passed:", len(port_passed))
print("Failed:", len(port_failed))


# 5. Average grades according to study time

print("\nAverage grades according to study time")

print(maths.groupby("studytime")["G3"].mean())


# 6. Relationship between absences and grades

print("\nCorrelation between absences and final grades")

print(maths[["absences", "G3"]].corr())


# 7. Visualization: Average grades comparison

subjects = ["Mathematics", "Portuguese"]

averages = [maths_average, portuguese_average]

plt.bar(subjects, averages)

plt.title("Average Student Performance")
plt.xlabel("Subjects")
plt.ylabel("Average Final Grade")

plt.show()


# 8. Visualization: Study time vs grades

study_grades = maths.groupby("studytime")["G3"].mean()

study_grades.plot(kind="bar")

plt.title("Study Time vs Average Mathematics Grade")
plt.xlabel("Study Time")
plt.ylabel("Average Grade")

plt.show()


# 9. Visualization: Absences vs grades

plt.scatter(maths["absences"], maths["G3"])

plt.title("Absences vs Mathematics Grades")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")

plt.show()
```
